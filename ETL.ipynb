```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("rezaghari/friends-series-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os
import re

df = pd.read_csv(os.path.join(path, "friends_episodes_v3.csv"), encoding='latin-1')

print(df.head())
```

```python
def limpar(texto):
    texto = texto.lower()
    texto = re.sub(r"[^\w\s]", "", texto)
    texto = texto.replace(" ", "_")
    return texto

linhas_prolog = []

for _, col in df.iterrows():
    titulo = limpar(col["Episode_Title"])
    temporada = col["Season"]
    num_ep = col["Episode Number"]
    ano = col["Year_of_prod"]
    duracao = col["Duration"]
    diretor = limpar(col["Director"])
    nota = col["Stars"]
    votos = col["Votes"]

    predicado = f"episodio({titulo}, {temporada}, {num_ep}, {ano}, {duracao}, {diretor}, {nota}, {votos})."
    
    linhas_prolog.append(predicado)

linhas_prolog
```

```python
with open("friends.pl", "w", encoding="utf-8") as f:
    for linha in linhas_prolog:
        f.write(linha + "\n")
```
